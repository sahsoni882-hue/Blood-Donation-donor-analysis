```python
# ============================================
# BLOOD DONATION & DONOR ANALYSIS DASHBOARD
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Blood Donation & Donor Analysis Dashboard")
print("System Ready!")
```

```python
np.random.seed(42)

donor_ids = [f"D{1001+i}" for i in range(200)]

first_names = [
    "Aarav", "Vivaan", "Aditya", "Arjun", "Rahul",
    "Rohan", "Vikram", "Karan", "Aman", "Nikhil",
    "Ananya", "Diya", "Ishita", "Priya", "Sneha",
    "Kavya", "Neha", "Pooja", "Riya", "Simran"
]

last_names = [
    "Sharma", "Verma", "Gupta", "Singh", "Kumar",
    "Patel", "Mehta", "Jain", "Yadav", "Mishra"
]

names = [
    np.random.choice(first_names) + " " + np.random.choice(last_names)
    for _ in range(200)
]

ages = np.random.randint(18, 61, 200)

genders = np.random.choice(
    ["Male", "Female"], 200
)

blood_groups = np.random.choice(
    ["A+", "A-", "B+", "B-", "AB+", "AB-", "O+", "O-"],
    200
)

cities = np.random.choice(
    ["Delhi", "Mumbai", "Jaipur", "Lucknow", "Kolkata",
     "Pune", "Bhopal", "Indore", "Chandigarh", "Noida"],
    200
)

phone_numbers = [
    "9" + "".join(np.random.choice(list("0123456789"), 9))
    for _ in range(200)
]

donation_count = np.random.randint(1, 11, 200)

last_donation_year = np.random.randint(2021, 2026, 200)

availability = np.random.choice(
    ["Available", "Not Available"], 200
)

df = pd.DataFrame({
    "Donor_ID": donor_ids,
    "Name": names,
    "Age": ages,
    "Gender": genders,
    "Blood_Group": blood_groups,
    "City": cities,
    "Phone": phone_numbers,
    "Donation_Count": donation_count,
    "Last_Donation_Year": last_donation_year,
    "Availability": availability
})

display(df)
```

```python
age_summary = df["Age"].describe()

print("Age Statistics:")
print(age_summary)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=10, rwidth=0.85)

plt.title("Age Distribution of Donors")
plt.xlabel("Age")
plt.ylabel("Number of Donors")

plt.tight_layout()
plt.show()
```

```python
city_counts = df["City"].value_counts()

print("City-wise Donor Count:")
print(city_counts)
```

```python
plt.figure(figsize=(9,5))

plt.bar(city_counts.index, city_counts.values)

plt.title("City-wise Donor Count")
plt.xlabel("City")
plt.ylabel("Number of Donors")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
donation_summary = df["Donation_Count"].describe()

print("Donation Statistics:")
print(donation_summary)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Donation_Count"], bins=10, rwidth=0.85)

plt.title("Donor Donation Count Distribution")
plt.xlabel("Number of Donations")
plt.ylabel("Number of Donors")

plt.tight_layout()
plt.show()
```

```python
blood_group_counts = df["Blood_Group"].value_counts()

print("Blood Group-wise Donor Count:")
print(blood_group_counts)
```

```python
plt.figure(figsize=(8,5))

plt.bar(blood_group_counts.index, blood_group_counts.values)

plt.title("Blood Group-wise Donor Count")
plt.xlabel("Blood Group")
plt.ylabel("Number of Donors")

plt.tight_layout()
plt.show()
```

```python
availability_counts = df["Availability"].value_counts()

print("Donor Availability:")
print(availability_counts)
```

```python
plt.figure(figsize=(6,5))

plt.bar(
    availability_counts.index,
    availability_counts.values
)

plt.title("Donor Availability")
plt.xlabel("Availability")
plt.ylabel("Number of Donors")

plt.tight_layout()
plt.show()
```

```python
donor_id = input("Enter Donor ID: ").strip().upper()

result = df[df["Donor_ID"] == donor_id]

if not result.empty:
    print("🩸 DONOR RECORD FOUND")
    display(result)
else:
    print("❌ Donor ID not found.")
```

```python
blood_group = input("Enter Blood Group: ").strip().upper()
city = input("Enter City: ").strip().title()

result = df[
    (df["Blood_Group"] == blood_group) &
    (df["City"] == city)
]

if not result.empty:
    print("🩸 DONORS FOUND")
    display(result)
else:
    print("❌ No matching donors found.")
```

```python
available_donors = df[df["Availability"] == "Available"]

print("Available Donors:")
display(available_donors)
```

```python
print("========== BLOOD DONATION DASHBOARD SUMMARY ==========")

print("Total Donors:", len(df))
print("Available Donors:", (df["Availability"] == "Available").sum())
print("Not Available Donors:", (df["Availability"] == "Not Available").sum())
print("Average Age:", round(df["Age"].mean(), 2))
print("Average Donation Count:", round(df["Donation_Count"].mean(), 2))

print("\nMost Common Blood Group:")
print(df["Blood_Group"].value_counts().idxmax())

print("\nTop Donor City:")
print(df["City"].value_counts().idxmax())
```
